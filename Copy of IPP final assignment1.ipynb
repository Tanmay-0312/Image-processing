Write a Python code snippet to crop the central 200×200 pixel area of an image and display it using Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread('/content/scene.jpg')

# Check if image is loaded
if img is not None:

    # Get image height and width
    height, width = img.shape[:2]

    # Calculate the centre coordinates
    x = (width - 200) // 2
    y = (height - 200) // 2

    # Crop the central 200x200 pixel area
    cropped = img[y:y+200, x:x+200]

    # Convert BGR to RGB
    cropped = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

    # Display the cropped image
    plt.imshow(cropped)
    plt.title("Cropped Image (200x200)")
    plt.axis('off')
    plt.show()

else:
    print("Image not found!")
```

Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread('/content/flower.jpg')

# Check if image is loaded
if img is not None:

    # Convert BGR to RGB
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Convert RGB to HSV
    hsv = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)

    # Split HSV channels
    hue, saturation, value = cv2.split(hsv)

    # Display each channel separately
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(hue, cmap='gray')
    plt.title("Hue")
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(saturation, cmap='gray')
    plt.title("Saturation")
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(value, cmap='gray')
    plt.title("Value")
    plt.axis('off')

    plt.show()

else:
    print("Image not found!")
```

Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, "img1.jpg" and "img2.jpg".

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the first image
img1 = cv2.imread('/content/flower.jpg')

# Read the second image
img2 = cv2.imread('/content/scene.jpg')

# Convert both images into grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Save grayscale images
cv2.imwrite('/content/img1.jpg', gray1)
cv2.imwrite('/content/img2.jpg', gray2)

# Display grayscale images
print("Grayscale Image 1")
cv2_imshow(gray1)

print("Grayscale Image 2")
cv2_imshow(gray2)

print("Both grayscale images saved successfully!")
```

```python
import cv2
import numpy as np

# Read both grayscale images
img1 = cv2.imread('/content/img1.jpg', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('/content/img2.jpg', cv2.IMREAD_GRAYSCALE)

# Resize second image to match first image dimensions
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Calculate Mean Squared Error
mse = np.mean((img1.astype(float) - img2.astype(float)) ** 2)

# Print MSE
print("Mean Squared Error (MSE):", mse)
```
